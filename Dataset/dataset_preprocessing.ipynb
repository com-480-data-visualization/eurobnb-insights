```python
import pandas as pd
import os
```

```python
lisbonWeekdays = pd.read_csv('./Processed-Dataset/lisbon_weekdays.csv')
lisbonWeekends = pd.read_csv('./Processed-Dataset/lisbon_weekends.csv')
londonWeekdays = pd.read_csv('./Processed-Dataset/london_weekdays.csv')
londonWeekends = pd.read_csv('./Processed-Dataset/london_weekends.csv')
parisWeekdays = pd.read_csv('./Processed-Dataset/paris_weekdays.csv')
parisWeekends = pd.read_csv('./Processed-Dataset/paris_weekends.csv')
romeWeekdays = pd.read_csv('./Processed-Dataset/rome_weekdays.csv')
romeWeekends = pd.read_csv('./Processed-Dataset/rome_weekends.csv')


```

```python
# Remove 50% of the data from each dataset
lisbonWeekdays = lisbonWeekdays.sample(frac=0.5, random_state=1)
lisbonWeekends = lisbonWeekends.sample(frac=0.5, random_state=1)
londonWeekdays = londonWeekdays.sample(frac=0.05, random_state=1)
londonWeekends = londonWeekends.sample(frac=0.05, random_state=1)
parisWeekdays = parisWeekdays.sample(frac=0.5, random_state=1)
parisWeekends = parisWeekends.sample(frac=0.5, random_state=1)
romeWeekdays = romeWeekdays.sample(frac=0.5, random_state=1)
romeWeekends = romeWeekends.sample(frac=0.5, random_state=1)
```

```python
# Save the reduced datasets
# Make the directory if it doesn't exist
output_dir = './Reduced-Dataset'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
lisbonWeekdays.to_csv(os.path.join(output_dir, 'lisbon_weekdays.csv'), index=False)
lisbonWeekends.to_csv(os.path.join(output_dir, 'lisbon_weekends.csv'), index=False)
londonWeekdays.to_csv(os.path.join(output_dir, 'london_weekdays.csv'), index=False)
londonWeekends.to_csv(os.path.join(output_dir, 'london_weekends.csv'), index=False)
parisWeekdays.to_csv(os.path.join(output_dir, 'paris_weekdays.csv'), index=False)
parisWeekends.to_csv(os.path.join(output_dir, 'paris_weekends.csv'), index=False)
romeWeekdays.to_csv(os.path.join(output_dir, 'rome_weekdays.csv'), index=False)
romeWeekends.to_csv(os.path.join(output_dir, 'rome_weekends.csv'), index=False)
print("Reduced datasets saved in './Reduced-Dataset' directory.")
```
